#### Conditionals: If-statements

In this section, you will learn about If-statements. These are very useful, because they allow you to execute parts of your code based on whether a condition is met. Here is a short example:

<code>
if condition :
    bit of code
else :
    bit of code
</code>

As long as the statement after the keyword "if" evaluates to True, the indented code block will run. If the statement is False, all the indented parts will just be skipped when running the program.

We have also already mentioned "else", which you can use right after an if-statement. The code block after the
"else" keyword will run if the code within the if-statement did NOT run.

There is also "elif", which lets you check additional conditions before moving on to the "else" part of your statement.

<code>
if [condition1] :
    bit of code
elif [condition2] :
    bit of code
else :
    bit of code
</code>

```python
price = 42

if price > 100:
    print("expensive")

elif price > 50:
    print("moderate")

else:
    print("cheap") 
```

```python
name = "John"
if len(name) < 3:
    print("your name is short!")
```

#### Functions

Functions are the building blocks of almost every Python program.
They’re where the real action takes place.

You’ve already seen how to use several functions, including print() and len(). These are all __built-in functions__ because they come built into the Python language itself. You can also create __user-defined functions__ that perform specific tasks.

Functions break code into smaller chunks, and are great for tasks that a program uses repeatedly. Instead of writing the same code each time the program need to perform the task, just call the function.


In the past few chapters you used functions like print() and len() to display text and determine the length of a string.

```python
type(len)
```

You must __call__ the function to tell Python to actually execute it.

__Use parentheses to call the function.__

```python
len()
```

In this example, Python raises a TypeError when len() is called because len() expects __an argument__.

An __argument__ is a value that gets passed to the function as __input__. Some functions can be called with no arguments, and some can take as many arguments as you like. len() requires exactly one argument.

Finally, when a function is done executing, it __returns__ a value as __output__.

<div class="alert-info"> 
<ol>
<li> The function is <b>called</b>, and any arguments are passed to the function as input. </li>
<li> The function <b>executes</b>, and some action is performed with the
arguments. </li>
    <li> The function <b>returns</b>, and the original function call is replaced
with the return value. </li>
</ol>
</div>

```python
num_letters = len("emlyon")
print(num_letters)
```

First, len() is called with the argument "emlyon". The length of the string "emlyon" is calculated, which is the number 6. Then len() returns the number 6 and replaces the function call with the value.

something like this: num_letters = 6

Then Python assigns the value 6 to num_letters and continues executing any remaining lines of code in the program, which is print(num_letters).

#### Write Your Own Functions

As you write longer and more complex programs, you may find that you need to use the same few lines of code repeatedly. Or maybe you need to calculate the same formula with different values several times in your code.

In this section, you’ll learn how to define your own functions so that you can avoid repeating yourself when you need to reuse code.

Every function has two parts:
1. The __function signature__ defines the name of the function and any inputs it expects.
2. The __function body__ contains the code that runs every time the function is used.

An example:

```python
def multiply(x, y): # Function signature
    
    product = x * y # Function body

    return product # Return statement
```

It might seem odd to make a function for something as simple as the * operator. In fact, multiply is not a function you would probably write in a real-world scenario. But it makes a great first example for understanding how functions are created!

When Python reads a line beginning with the __def__ keyword, it creates a new function. The function is assigned to a variable with the same name as the function name.

<div class="alert-warning"> 
Since function names become variables, they must follow the same rules for variable names that you learned before.
</div>

You call a user-defined function just like any other function. Type the function name followed by a list of arguments in between parentheses.

For instance, to call multiply() with the argument 2 and 4, just type:

```python
multiply(2,4) 
```

#### Understand Scope in Python

When you assign a value to a variable, you are giving that value a name. Names are unique. For example, you can’t assign the same name to two different numbers.

```python
x = 2
x
```

```python
x = 3
x
```

When you assign 3 to x, you can no longer recall the value 2 with the name x which makes sense.

Now, try running the following script:

```python
x = 2

def func():
    x = 3
    print("Inside 'func', x has the value ",x)
    
func()

print("Outside 'func', x has the value ",x)
```

How does x still have the value 2 after calling func(), which changes the value of x to 3?

The answer is that the function func() has a __different scope__ than the code that exists outside of the function.

<div class="alert-info"> 
The function body has what is known as a <b>local scope</b>, with its own set of names available to it. Code outside of the function body is in the <b>global scope</b>.
</div>

Scopes have a hierarchy. For example, consider the following:

```python
x = 5

def func():
    y = x + 5
    
    print("Inside 'func', y has the value ",y)
    
func()

print("Outside 'func', x has the value ",x)
```

```python
print("Outside 'func', y has the value ",y)
```

The variable y is in the local scope of func(). When Python executes the line y = x + 5, it looks for the variables x in the local scope. However, x DOESN'T exist there, so it moves up to the global scope and looks for x there.

#### Loops:

Since we are programmers we don't want to work hard but smart, so we need something to help us work smart and that's where loops come into play. 

For example imagine your boss comes up to you and wants you to print out the numbers from 1 to 10. 
Well without knowing about loops you will probably write "print(<your_number>)" ten times with each number, but this is really slow and monotonous work, so we can introduce loops to take care of such chores for us!


Loops help us with tasks that sometimes are really similar, like the printing numbers example, or more complex, for example when we include conditionals or another for loop inside our existing for loop.

But don't worry, we will explain how all that work in the sections below!

__For loops structure:__

<code>
for [every_element] in [my_sequence]:
   do something
</code>

The structure might be confusing at first, but can be understood easily.

__every_element__ is the variable that gets created using the for loop and that variable will iterate over the given sequence __my_sequence__ from start to finish. So depending on how long that sequence is, the loop will "do something" for that many times.

As an example printing the numbers would be:

```python
for i in range(1,11):
    print(i)
```

where __i__ is the variable that iterates over our sequence, in this case a sequence of the numbers from 1 to 10 using the build-in function range().

range(1, 11) is the range of numbers 1, 2, 3, untill 10. The first argument is the starting number, and the second argument is the endpoint, which is not included in the range.

<div class="alert-info">
range() can be very helpful when defining a for-loop. It represents a sequence of numbers that you can then use to iterate over, and it takes three parameters: start, stop, and (optionally) step. Intuitively, the start and stop parameters define where your range begins and where it ends. If you do not define a step parameter, by default, your loop will move through the range with the step size 1.
<div>

__break, continue and pass__

```python
for i in range(1,11):
    if i == 5:
        continue
        print('we are here')
    
    print(i)
```

```python
for i in range(1,11):
    if i == 5:
        break
        print('we are here')
    
    print(i)
```

```python
for i in range(1,11):
    if i == 5:
        pass
        print('we are here')
    
    print(i)
```

__While Loops structure:__

While Loop might be used for some more uncertain tasks where we only know when to stop, but not how many times a certain task needs to be done.

Again structure wise the While Loops can be understood like:
 
<code>
while [condition]:
   do whatever
</code>

So before "starting" the while loop, the __condition__ needs to be checked and only if it is True the while loop will actually "do whatever". Imagine __condition__ as a conditional or just a boolean variable that would get set to False if a certain "event" would happen inside the while loop.

To use the same printing numbers example a while loop would look like this:

```python
counter = 1
while counter < 11:
    print(counter)
    counter = counter + 1
```

So the conditional is True as long as the __counter__ is smaller than 11 and once it is not the while loop will stop.

This is an example where both loops differ, since the for loop just iterates over a given sequence and is finite in that sense and the while loop actually could run forever (if you are not careful) if the conditional never gets False.

Another difference is the variable in use, since the for loop actually creates a variable to iterate over the sequence the while loops doesn't and that why we have to create one for ourselves.
